```python
import pandas as pd

# =====================================================
# EXPERIMENT NO: 3
# DATA CLEANING - TOURISM DATASET
# =====================================================

# Load the dataset
df = pd.read_csv("tourism_dataset_200.csv")

# =====================================================
# 1. DATASET INFORMATION
# =====================================================

print("===== DATASET INFORMATION =====")
df.info()

print("\n===== FIRST 5 RECORDS =====")
print(df.head())


# =====================================================
# 2. CHECK MISSING VALUES
# =====================================================

print("\n===== MISSING VALUES IN EACH COLUMN =====")
print(df.isnull().sum())

print("\n===== TOTAL MISSING VALUES =====")
print(df.isnull().sum().sum())

print("\n===== ANY MISSING VALUES? =====")
print(df.isnull().values.any())

print("\n===== NON-MISSING VALUES COUNT =====")
print(df.notnull().sum())


# =====================================================
# 3. FILL MISSING VALUES WITH CONSTANT VALUE
# =====================================================

df_constant = df.fillna(0)

print("\n===== FILL MISSING VALUES WITH CONSTANT (0) =====")
print(df_constant.head())


# =====================================================
# 4. FILL MISSING VALUES USING MEAN
# =====================================================

df_mean = df.copy()

df_mean["Rating"] = df_mean["Rating"].fillna(
    df_mean["Rating"].mean()
)

print("\n===== FILL MISSING VALUES USING MEAN =====")
print(df_mean.head())


# =====================================================
# 5. FILL MISSING VALUES USING MEDIAN
# =====================================================

df_median = df.copy()

df_median["Average_Duration_Days"] = \
    df_median["Average_Duration_Days"].fillna(
        df_median["Average_Duration_Days"].median()
    )

print("\n===== FILL MISSING VALUES USING MEDIAN =====")
print(df_median.head())


# =====================================================
# 6. FILL MISSING VALUES USING MODE
# =====================================================

df_mode = df.copy()

df_mode["Country"] = df_mode["Country"].fillna(
    df_mode["Country"].mode()[0]
)

print("\n===== FILL MISSING VALUES USING MODE =====")
print(df_mode.head())


# =====================================================
# 7. FORWARD FILL
# =====================================================

df_ffill = df.ffill()

print("\n===== FORWARD FILL =====")
print(df_ffill.head())


# =====================================================
# 8. BACKWARD FILL
# =====================================================

df_bfill = df.bfill()

print("\n===== BACKWARD FILL =====")
print(df_bfill.head())


# =====================================================
# 9. DROP ROWS WITH MISSING VALUES
# =====================================================

df_drop_rows = df.dropna()

print("\n===== DROP ROWS WITH MISSING VALUES =====")
print(df_drop_rows.head())


# =====================================================
# 10. DROP COLUMNS WITH MISSING VALUES
# =====================================================

df_drop_cols = df.dropna(axis=1)

print("\n===== DROP COLUMNS WITH MISSING VALUES =====")
print(df_drop_cols.head())


# =====================================================
# 11. DROP ROWS HAVING ANY MISSING VALUE
# =====================================================

df_any = df.dropna(how="any")

print("\n===== DROP HOW = 'ANY' =====")
print(df_any.head())


# =====================================================
# 12. DROP ROWS WHERE ALL VALUES ARE MISSING
# =====================================================

df_all = df.dropna(how="all")

print("\n===== DROP HOW = 'ALL' =====")
print(df_all.head())


# =====================================================
# 13. KEEP ROWS WITH AT LEAST 10 NON-NULL VALUES
# =====================================================

df_thresh = df.dropna(thresh=10)

print("\n===== THRESHOLD = 10 =====")
print(df_thresh.head())


# =====================================================
# 14. DROP ROWS BASED ON SPECIFIC COLUMNS
# =====================================================

df_subset = df.dropna(
    subset=["Country", "Rating"]
)

print("\n===== DROP MISSING VALUES FROM COUNTRY AND RATING =====")
print(df_subset.head())


# =====================================================
# 15. DROP MISSING VALUES PERMANENTLY
# =====================================================

df_cleaned = df.copy()

df_cleaned.dropna(inplace=True)

print("\n===== AFTER DROPNA(inplace=True) =====")
print(df_cleaned.head())


# =====================================================
# FINAL DATASET INFORMATION
# =====================================================

print("\n===== FINAL DATASET INFORMATION =====")
df_cleaned.info()

print("\n===== DATA CLEANING COMPLETED SUCCESSFULLY =====")
```
