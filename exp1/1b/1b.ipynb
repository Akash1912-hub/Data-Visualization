```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# CRISP-DM FRAMEWORK
# ============================================================

# 1. BUSINESS UNDERSTANDING
# Objective:
# Analyze tourism destinations, travel preferences,
# budgets, ratings, and visitor patterns.

# 2. DATA UNDERSTANDING
df = pd.read_csv("tourism_dataset_200.csv")
print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
print(df.isnull().sum())
print(df.duplicated().sum())

# 3. DATA PREPARATION
df.drop_duplicates(inplace=True)
df['Rating'] = df['Rating'].fillna(df['Rating'].median())
df['Estimated_Budget_USD'] = df['Estimated_Budget_USD'].fillna(
    df['Estimated_Budget_USD'].median()
)
df['Annual_Visitors'] = df['Annual_Visitors'].fillna(
    df['Annual_Visitors'].median()
)

# Derived feature
df['Budget_Per_Day_USD'] = (
    df['Estimated_Budget_USD'] / df['Average_Duration_Days']
).round(2)

# 4. MODELING (EDA)
df['Tourism_Type'].value_counts().plot(kind='bar')
plt.title("Tourism Type Distribution")
plt.xlabel("Tourism Type")
plt.ylabel("Count")
plt.show()

df['Continent'].value_counts().plot(kind='bar')
plt.title("Tourists Destinations by Continent")
plt.xlabel("Continent")
plt.ylabel("Count")
plt.show()

df['Best_Season'].value_counts().plot(kind='bar')
plt.title("Best Season for Tourism")
plt.xlabel("Season")
plt.ylabel("Count")
plt.show()

df.groupby('Country')['Estimated_Budget_USD'].mean().sort_values(
    ascending=False
).head(10).plot(kind='bar')
plt.title("Top 10 Countries by Average Estimated Budget")
plt.xlabel("Country")
plt.ylabel("Average Budget (USD)")
plt.show()

corr = df.select_dtypes(include='number').corr()
plt.figure(figsize=(10, 8))
plt.imshow(corr)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Correlation Heatmap")
plt.show()

# 5. EVALUATION
print("Tourism insights generated successfully.")

# 6. DEPLOYMENT
print("Insights can be deployed through tourism dashboards.")


# SEMMA FRAMEWORK
# ============================================================

# SAMPLE
df = pd.read_csv("tourism_dataset_200.csv")

# EXPLORE
print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())

# MODIFY
df.drop_duplicates(inplace=True)
df['Rating'] = df['Rating'].fillna(df['Rating'].median())
df['Estimated_Budget_USD'] = df['Estimated_Budget_USD'].fillna(
    df['Estimated_Budget_USD'].median()
)
df['Annual_Visitors'] = df['Annual_Visitors'].fillna(
    df['Annual_Visitors'].median()
)
df['Budget_Per_Day_USD'] = (
    df['Estimated_Budget_USD'] / df['Average_Duration_Days']
).round(2)

# MODEL
df['Tourism_Type'].value_counts().plot(kind='bar')
plt.show()

df['Continent'].value_counts().plot(kind='bar')
plt.show()

df['Best_Season'].value_counts().plot(kind='bar')
plt.show()

df.groupby('Country')['Estimated_Budget_USD'].mean().sort_values(
    ascending=False
).head(10).plot(kind='bar')
plt.show()

corr = df.select_dtypes(include='number').corr()
plt.figure(figsize=(10, 8))
plt.imshow(corr)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.show()

# ASSESS
print("Tourism patterns and trends successfully identified.")


# KDD FRAMEWORK
# ============================================================

# SELECTION
df = pd.read_csv("tourism_dataset_200.csv")

# PREPROCESSING
print(df.info())
print(df.isnull().sum())
print(df.duplicated().sum())

df.drop_duplicates(inplace=True)
df['Rating'] = df['Rating'].fillna(df['Rating'].median())
df['Estimated_Budget_USD'] = df['Estimated_Budget_USD'].fillna(
    df['Estimated_Budget_USD'].median()
)
df['Annual_Visitors'] = df['Annual_Visitors'].fillna(
    df['Annual_Visitors'].median()
)

# TRANSFORMATION
df['Budget_Per_Day_USD'] = (
    df['Estimated_Budget_USD'] / df['Average_Duration_Days']
).round(2)

# DATA MINING
df['Tourism_Type'].value_counts().plot(kind='bar')
plt.show()

df['Continent'].value_counts().plot(kind='bar')
plt.show()

df['Best_Season'].value_counts().plot(kind='bar')
plt.show()

df.groupby('Country')['Estimated_Budget_USD'].mean().sort_values(
    ascending=False
).head(10).plot(kind='bar')
plt.show()

corr = df.select_dtypes(include='number').corr()
plt.figure(figsize=(10, 8))
plt.imshow(corr)
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.show()

# INTERPRETATION
print("Knowledge discovered from the tourism dataset has been interpreted to support tourism planning and business decisions.")
```
