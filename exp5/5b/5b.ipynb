```python
import pandas as pd
import matplotlib.pyplot as plt

# =====================================================
# EXPERIMENT NO: 5
# DATA VISUALIZATION USING MATPLOTLIB
# TOURISM DATASET
# =====================================================


# -----------------------------------------------------
# LOAD DATASET
# -----------------------------------------------------

df = pd.read_csv("tourism_dataset_200.csv")

print("===== TOURISM DATASET LOADED SUCCESSFULLY =====")


# =====================================================
# ALGORITHM 1: LINE CHART
# Number of Destinations by Continent
# =====================================================

continent_counts = df["Continent"].value_counts()

plt.figure(figsize=(10, 5))

plt.plot(
    continent_counts.index,
    continent_counts.values,
    marker="o"
)

plt.title("Number of Tourism Destinations by Continent")

plt.xlabel("Continent")

plt.ylabel("Number of Destinations")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()


# =====================================================
# ALGORITHM 2: BAR CHART
# Number of Destinations by Tourism Type
# =====================================================

tourism_type_counts = df["Tourism_Type"].value_counts()

plt.figure(figsize=(10, 5))

plt.bar(
    tourism_type_counts.index,
    tourism_type_counts.values
)

plt.title("Number of Destinations by Tourism Type")

plt.xlabel("Tourism Type")

plt.ylabel("Number of Destinations")

plt.xticks(rotation=45)

plt.show()


# =====================================================
# ALGORITHM 3: PIE CHART
# Accommodation Type Distribution
# =====================================================

accommodation_counts = df["Accommodation_Type"].value_counts()

plt.figure(figsize=(8, 6))

plt.pie(
    accommodation_counts.values,
    labels=accommodation_counts.index,
    autopct="%1.1f%%"
)

plt.title("Accommodation Type Distribution")

plt.show()


# =====================================================
# ALGORITHM 4: HISTOGRAM
# Estimated Budget Distribution
# =====================================================

plt.figure(figsize=(8, 5))

plt.hist(
    df["Estimated_Budget_USD"],
    bins=20
)

plt.title("Distribution of Estimated Tourism Budget")

plt.xlabel("Estimated Budget (USD)")

plt.ylabel("Frequency")

plt.show()


# =====================================================
# ALGORITHM 5: SCATTER PLOT
# Average Duration vs Estimated Budget
# =====================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Average_Duration_Days"],
    df["Estimated_Budget_USD"],
    alpha=0.5
)

plt.title(
    "Average Duration vs Estimated Budget"
)

plt.xlabel("Average Duration (Days)")

plt.ylabel("Estimated Budget (USD)")

plt.show()


# =====================================================
# ALGORITHM 6: BOX PLOT
# Estimated Budget
# =====================================================

plt.figure(figsize=(7, 5))

plt.boxplot(
    df["Estimated_Budget_USD"].dropna()
)

plt.title(
    "Estimated Tourism Budget Box Plot"
)

plt.ylabel("Estimated Budget (USD)")

plt.show()


# =====================================================
# ALGORITHM 7: MULTIPLE LINE CHART
# Average Budget and Average Duration by Tourism Type
# =====================================================

tourism_stats = df.groupby(
    "Tourism_Type"
).agg({
    "Estimated_Budget_USD": "mean",
    "Average_Duration_Days": "mean"
})

plt.figure(figsize=(10, 5))

plt.plot(
    tourism_stats.index,
    tourism_stats["Estimated_Budget_USD"],
    marker="o",
    label="Average Budget"
)

plt.plot(
    tourism_stats.index,
    tourism_stats["Average_Duration_Days"],
    marker="o",
    label="Average Duration"
)

plt.title(
    "Average Budget and Duration by Tourism Type"
)

plt.xlabel("Tourism Type")

plt.ylabel("Average Value")

plt.xticks(rotation=45)

plt.legend()

plt.grid(True)

plt.show()


# =====================================================
# END OF PROGRAM
# =====================================================

print("===== DATA VISUALIZATION COMPLETED SUCCESSFULLY =====")
```
