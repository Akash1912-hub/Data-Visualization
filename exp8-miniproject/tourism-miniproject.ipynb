```python
import pandas as pd

file_path = "tourism_dataset_200.csv"
df = pd.read_csv(file_path)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
df.head()
```

## Data Cleaning and Transformation

```python
# Make a copy
tourism = df.copy()

# Handle missing values using suitable defaults
tourism["Destination"] = tourism["Destination"].fillna("Unknown")
tourism["Country"] = tourism["Country"].fillna("Unknown")
tourism["Tourism_Type"] = tourism["Tourism_Type"].fillna("Unknown")

# Remove duplicate records
tourism.drop_duplicates(inplace=True)

# Create useful derived columns
tourism["Budget_Per_Day_USD"] = (
    tourism["Estimated_Budget_USD"] /
    tourism["Average_Duration_Days"]
).round(2)

def visitor_category(x):
    if x >= 500000:
        return "High"
    elif x >= 200000:
        return "Medium"
    return "Low"

tourism["Visitor_Category"] = tourism["Annual_Visitors"].apply(visitor_category)

# Standardize column names
tourism.columns = (
    tourism.columns.str.strip().str.replace(" ", "_", regex=False)
)

print("Original rows:", len(df))
print("Cleaned rows:", len(tourism))
print("Columns:", len(tourism.columns))
print("\nMissing values:")
print(tourism.isnull().sum()[tourism.isnull().sum() > 0])
tourism.head()
```
